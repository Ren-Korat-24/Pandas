```python
n=int(input("Enter a number: "))
fact=1

for i in range(1,n+1):
    fact*=i

print(f"The factorial of {n} is {fact}")
```

```python
num=int(input("Enter a number: "))

if num<2:
    print("not prime")
else:
    for i in range(2,num):
        if num%i==0:
            print("not prime")
            break
    else:
        print("Prime")
```

```python
data={'name':['aice','bob','chgarlie'],
      'maths':[80,90,70],
      'english':[85,95,75],
      'science':[90,80,85]}

df=pd.DataFrame(data)

df['total_marks']=df[['maths','english','science']].sum(axis=1)
df['average_marks']=df[['maths','english','science']].mean(axis=1)

df
```
