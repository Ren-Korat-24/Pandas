Row Selection, Addition, and Deletion

```python
ans=({'design':pd.Series([1,2,3],index=['a','b','c']),'development':pd.Series([4,5,6],index=['a','b','c'])})

rs=pd.DataFrame(ans)

rs['skills']=pd.Series([20,30,60],index=['a','b','c'])

#addition
print(rs['skills']+ rs['development'])

#deletion
# del rs['design']
print(rs)

#selection single row   
print(rs.loc['a'])

#select single column
print(rs.loc['a':'b','development'])

#Select Multiple rows
print(rs.loc[['a','c']])

#Select Multiple columns 
print(rs.loc[:,['development','skills']])

#Select row range 
print(rs.loc["a":"c"])

#select column range 
print(rs.loc[:,'design':'development'])

#select alternative row 
print(rs.loc['a':'c':1])

#select alternative column
print(rs.loc[:,'development':'skills':2])

#using condioon0
rs.loc[rs['design'] >= 2]
```
