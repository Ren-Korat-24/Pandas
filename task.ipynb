4.Create a Series [1, None, 3, None, 5] and fill missing values with 0.

5.Find the sum, mean, maximum, and minimum of a Series [5, 10, 15, 20, 25]

```python
import pandas as pd
import numpy as np

df=np.array([5, 10, 15, 20, 25])
sr=pd.Series(df)
f1=sr.agg([np.sum,np.mean,np.max,np.min])
print(f1)
```

9.Group the DataFrame by Grade and find the average Marks.

```python
import pandas as pd
import numpy as np

data = {'Name': ['sachin', 'raj', 'virat', 'dhoni'],   
   'Percentage': [82, 98, 91, 87],   
   'grade': ['A', 'A+', 'A+', 'A']}  

df=pd.DataFrame(data)

gb=df.groupby('grade')

print(gb['Percentage'].agg(np.average))
```

11.Create two Series [1,2,3,4,5] and [5,4,3,2,1]. Perform element-wise multiplication

```python
import pandas as pd   
import numpy as np   

s1=pd.Series([1,2,3,4,5])
s2=pd.Series([5,4,3,2,1])
# mul=s1*s2
multiplication=s1.multiply(s2)
print(multiplication)
```

13.Create a DataFrame of 3 students with marks in 3 subjects. Add total and average columns

```python
data={'name':['aice','bob','chgarlie'],
      'maths':[80,90,70],
      'english':[85,95,75],
      'science':[90,80,85]}

df=pd.DataFrame(data)

df['total_marks']=df[['maths','english','science']].sum(axis=1)
df['average_marks']=df[['maths','english','science']].mean(axis=1)

df
```
