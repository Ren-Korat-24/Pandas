Series.shape It returns a tuple of shape of the data. 


```python
dt=np.array([1,2,3,4]) 
sr=pd.Series(dt,index=['alpha','beta','gama','delta']) 
print(sr) 
print(sr.shape) 
```

Series.dtype It returns the data type of the data. 


Series.empty It returns True if Series object is empty, otherwise returns false. 


Series.hasnans It returns True if there are any NaN values, otherwise returns false. 


Series.nbytes It returns the number of bytes in the data. 


Series.ndim It returns the number of dimensions in the data. 


```python
dt=np.array([1,2,3,4,np.nan]) 
sr=pd.Series(dt,index=['alpha','beta','gama','delta','Epsilon']) 
# print(sr) 
# print(sr.empty) 
# print(sr.hasnans) 
# print(len(sr)) 

print(sr.count) 
```

```python
dt=np.array([
    [1,2,3,4,np.nan],
    [5,6,7,8,9]]) 

sr=pd.DataFrame(dt,
            index=['alpha','beta'],
            columns=['a','b','c','d','e']) 
# print(sr)

print(sr.nbytes)  #total bytes consumed by the elements of the Series
print(sr.ndim) 
```
